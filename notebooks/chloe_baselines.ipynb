# Chloe — Baselines & course models (Sections 3–4)
Model code goes in `src/a2/models.py`; this notebook runs and compares on `valid`.

```python
from a2.data import load_dataset, split, targets, FIT_LABELS
from a2.evaluate import report, results_table
from a2.models import MajorityClass, ItemMajority, UserMajority

rows = load_dataset()
train, valid, test = split(rows)
len(train), len(valid), len(test)
```

```python
results = []
for name, model in [("Majority class", MajorityClass()), ("Item majority", ItemMajority()), ("User majority", UserMajority())]:
    model.fit(train)
    results.append(report(name, targets(valid), model.predict(valid)))
results_table(results)
```
