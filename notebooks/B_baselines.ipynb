# B — Baselines and course models (Sections 3–4)
_Model code goes in `src/a2/models.py`; this notebook runs and compares._

```python
from a2.data import load_dataset, split, targets
from a2.evaluate import report, results_table
from a2.models import GlobalMean, ItemMean, UserMean

rows = load_dataset()
train, valid, test = split(rows)
len(train), len(valid), len(test)
```

```python
results = []
for name, model in [("Global mean", GlobalMean()), ("Item mean", ItemMean()), ("User mean", UserMean())]:
    model.fit(train)
    results.append(report(name, targets(valid), model.predict(valid)))
results_table(results)
```
